```python
# 1. 设置镜像环境变量
!export HF_ENDPOINT=https://hf-mirror.com

# 2. 运行一行 Python 代码直接下载 GraphCodeBERT 到你指定的路径
from huggingface_hub import snapshot_download
try:
    path = snapshot_download(
        repo_id='microsoft/graphcodebert-base',
        cache_dir='/root/autodl-tmp/water/hf_cache',
        resume_download=True,
        ignore_patterns=['*.msgpack', '*.h5', '*.ot'] # 忽略不需要的格式，加速下载
    )
    print(f'\n✨ GraphCodeBERT 下载成功！路径: {path}')
except Exception as e:
    print(f'\n❌ 下载失败: {e}')
```
